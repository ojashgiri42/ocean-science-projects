```python
#Satellite-Based Assessment of Chlorophyll-a Variability in the Arabian Sea Using MODIS-Aqua Ocean Color Data

import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from io import StringIO

# NOAA MODIS-Aqua 8-day chlorophyll-a dataset
base_url = ( "https://oceanwatch.pifsc.noaa.gov/erddap/griddap/" "aqua_chla_8d_2018_0.csv")

# Defining Arabian Sea study area
lat_min = 5 
lat_max = 25 
lon_min = 45 
lon_max = 75 
date = "2021-06-01T00:00:00Z"

# Requesting subset from ERDDAP
query = ( 
    "?chlor_a"
    f"[({date})]"
    f"[({lat_min}):5:({lat_max})]"
    f"[({lon_min}):5:({lon_max})]"
)
url = base_url + query

print("Downloading Arabian Sea chlorophyll data...")

response = requests.get( url, timeout=120 )
response.raise_for_status()

print("Download successful!")

df = pd.read_csv( StringIO(response.text), skiprows=[1] )

print("\nData preview:")
print(df.head())
print("\nNumber of observations:", len(df))

# Removing missing and invalid values
df["chlor_a"] = pd.to_numeric( df["chlor_a"], errors="coerce" )
df["latitude"] = pd.to_numeric(  df["latitude"], errors="coerce" )
df["longitude"] = pd.to_numeric( df["longitude"], errors="coerce" )
df = df.dropna( subset=[ "chlor_a", "latitude", "longitude" ] )


# Making Chlorophyll positive for logarithmic scale
df = df[ df["chlor_a"] > 0 ]
print( "\nValid chlorophyll observations:", len(df) )

print("\nChlorophyll-a statistics:")
print( df["chlor_a"].describe() )

# Plotting map
fig = plt.figure( figsize=(12, 8) )
ax = plt.axes( projection=ccrs.PlateCarree() )

# Seting Arabian Sea extent
ax.set_extent( [ lon_min, lon_max, lat_min, lat_max ], crs=ccrs.PlateCarree() )

# Adding features
ax.add_feature( cfeature.LAND, facecolor="lightgray", zorder=2 )
ax.add_feature( cfeature.COASTLINE, linewidth=1.0, zorder=3 )
ax.add_feature( cfeature.BORDERS, linewidth=0.5, zorder=3 )

# Plotting chlorophyll-a

scatter = ax.scatter( df["longitude"],  df["latitude"], c=df["chlor_a"], s=12, cmap="viridis",
norm=colors.LogNorm(  vmin=0.05,  vmax=10 ), transform=ccrs.PlateCarree(), edgecolors="none" )


gl = ax.gridlines(  draw_labels=True, linestyle="--", alpha=0.5 )
gl.top_labels = False 
gl.right_labels = False
cbar = plt.colorbar( scatter,  ax=ax, shrink=0.8, pad=0.04 )
cbar.set_label( "Chlorophyll-a (mg m⁻³)", fontsize=11 )

plt.title( "MODIS-Aqua Chlorophyll-a Distribution\n" "Arabian Sea — June 2021",  fontsize=15, pad=12 )

plt.savefig( "../figures/arabian_sea_chlorophyll_june2021.png",  dpi=300, bbox_inches="tight" )
plt.show()
```
